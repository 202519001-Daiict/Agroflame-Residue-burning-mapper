# Notebook 08 — Crop Data Cleaning (Punjab + Haryana)

**Input:** Punjab and Haryana crop production XLS files  
**Output:** `Data/Processed/crop_production_clean.csv`

```python
import pandas as pd
```

```python
# ── Load both XLS files ───────────────────────────────────────────────
punjab_raw = pd.read_html(
    r"C:\Users\DSSON\Documents\M.Sc. Sem 2\Residue_Project\Data\Raw\Crop\Punjab Statistical Abstract (Crop Area + Production)\Punjab_CropProduction.xls"
)[0]
haryana_raw = pd.read_excel(r"C:\Users\DSSON\Documents\M.Sc. Sem 2\Residue_Project\Data\Raw\Crop\Haryana Statistical Abstract (Crop Area + Production)\Haryana_crop_raw(2022-23).xlsx",engine='openpyxl')
```

```python
# ── Flatten MultiIndex columns ────────────────────────────────────────
def flatten_cols(df):
    df.columns = ['_'.join(str(l) for l in col if str(l) != 'nan').strip('_') for col in df.columns]
    return df

punjab_raw  = flatten_cols(punjab_raw)
haryana_raw = flatten_cols(haryana_raw)

# Check column names (update col_map below if Haryana differs)
print(punjab_raw.columns.tolist())
print(haryana_raw.columns.tolist())
```

```python
# ── Rename and select columns ─────────────────────────────────────────
# Update keys if your Haryana column names differ from Punjab
col_map = {
    'State_State_State'               : 'state',
    'District_District_District'      : 'district',
    'Year_Year_Year'                  : 'year',
    'Rice_Kharif_Production (Tonnes)' : 'rice_prod',
    'Wheat_Rabi_Production (Tonnes)'  : 'wheat_prod',
}

keep = ['state', 'district', 'year', 'rice_prod', 'wheat_prod']

punjab  = punjab_raw.rename(columns=col_map)[keep].dropna(subset=['district', 'year'])
haryana = haryana_raw.rename(columns=col_map)[keep].dropna(subset=['district', 'year'])
```

```python
# ── Reshape wide → long and combine ──────────────────────────────────
def to_long(df):
    long = df.melt(id_vars=['state', 'district', 'year'], var_name='crop', value_name='production_t')
    long['crop'] = long['crop'].map({'rice_prod': 'Rice', 'wheat_prod': 'Wheat'})
    return long.dropna(subset=['production_t']).query('production_t > 0')

crop_data = pd.concat([to_long(punjab), to_long(haryana)], ignore_index=True)

print(f"Shape: {crop_data.shape}")
print(crop_data['state'].value_counts())
print(crop_data.head())
```

```python
# ── Save ──────────────────────────────────────────────────────────────
crop_data.to_csv('../Data/Processed/crop_production_clean.csv', index=False)
print(f"✅ Saved {len(crop_data)} rows → crop_production_clean.csv")
```
