# Notebook 08 — Crop Data Cleaning (Punjab + Haryana)

**Input:** `Punjab_CropProduction.xls`, `Haryana_crop_raw_2022-23_.xlsx`  
**Output:** `Data/Processed/crop_production_clean.csv`

### Key structural differences between the two source files

| | Punjab | Haryana |
|---|---|---|
| Format | HTML table saved as `.xls` | True `.xlsx` |
| Read with | `pd.read_html()` | `pd.read_excel(engine='openpyxl', header=3)` |
| Header rows | 3-level MultiIndex | Single header on row 4 (index 3) |
| State column | Present | Absent — added manually |
| Year column | Present (multi-year) | Absent — added as `'2022-23'` |
| Production units | Tonnes | **000 Tonnes** → multiply × 1000 |

```python
import pandas as pd
```

```python
# ── Punjab ────────────────────────────────────────────────────────────
# The file is an HTML table saved with a .xls extension.
# pd.read_excel(engine='xlrd') raises "Expected BOF record" on it.
# pd.read_html() is the correct loader.

PUNJAB_PATH  = r"C:\Users\DSSON\Documents\M.Sc. Sem 2\Residue_Project\Data\Raw\Crop\Punjab Statistical Abstract (Crop Area + Production)\Punjab_CropProduction.xls"
HARYANA_PATH = r"C:\Users\DSSON\Documents\M.Sc. Sem 2\Residue_Project\Data\Raw\Crop\Haryana Statistical Abstract (Crop Area + Production)\Haryana_crop_raw(2022-23).xlsx"

punjab_raw = pd.read_html(PUNJAB_PATH, header=[0, 1, 2])[0]

# Flatten 3-level MultiIndex → single string column names
punjab_raw.columns = [
    '_'.join(str(l) for l in col if str(l).lower() != 'nan').strip('_')
    for col in punjab_raw.columns
]

print(punjab_raw.columns.tolist())
punjab_raw.head(3)
```

```python
# ── Haryana ───────────────────────────────────────────────────────────
# Real .xlsx.  Rows 0-2 are title / source / crop group labels.
# Row 3 (0-indexed) contains the actual column headers → header=3.
# Units in this file are 000 ha / 000 Tonnes, NOT raw hectares/tonnes.

haryana_raw = pd.read_excel(HARYANA_PATH, engine='openpyxl', header=3)

# Assign clean column names by position (safe regardless of newlines / unicode in headers)
haryana_raw.columns = [
    'district',
    'rice_area_000ha', 'rice_prod_000t', 'rice_yield_kgha',
    'wheat_area_000ha', 'wheat_prod_000t', 'wheat_yield_kgha',
    'total_area', 'total_prod', 'residue'
]

print(haryana_raw.columns.tolist())
haryana_raw.head(3)
```

```python
# ── Select & rename Punjab columns ───────────────────────────────────
col_map_punjab = {
    'State_State_State'               : 'state',
    'District_District_District'      : 'district',
    'Year_Year_Year'                  : 'year',
    'Rice_Kharif_Production (Tonnes)' : 'rice_prod',
    'Wheat_Rabi_Production (Tonnes)'  : 'wheat_prod',
}

punjab = (
    punjab_raw
    .rename(columns=col_map_punjab)
    [['state', 'district', 'year', 'rice_prod', 'wheat_prod']]
    .dropna(subset=['district', 'year'])
)

print(f"Punjab: {punjab.shape}")
punjab.head(3)
```

```python
# ── Clean Haryana ─────────────────────────────────────────────────────
# Drop the aggregate total row and any blank district rows
haryana = haryana_raw[
    haryana_raw['district'].notna() &
    (haryana_raw['district'] != 'HARYANA TOTAL')
].copy()

# Add state and year (absent from file)
haryana['state'] = 'Haryana'
haryana['year']  = '2022-23'

# Convert 000 Tonnes → Tonnes (to match Punjab's unit)
haryana['rice_prod']  = pd.to_numeric(haryana['rice_prod_000t'],  errors='coerce') * 1000
haryana['wheat_prod'] = pd.to_numeric(haryana['wheat_prod_000t'], errors='coerce') * 1000

haryana = haryana[['state', 'district', 'year', 'rice_prod', 'wheat_prod']]

print(f"Haryana: {haryana.shape}")
haryana.head(3)
```

```python
# ── Reshape wide → long and combine ──────────────────────────────────
def to_long(df):
    long = df.melt(
        id_vars=['state', 'district', 'year'],
        var_name='crop',
        value_name='production_t'
    )
    long['crop'] = long['crop'].map({'rice_prod': 'Rice', 'wheat_prod': 'Wheat'})
    return long.dropna(subset=['production_t']).query('production_t > 0')

crop_data = pd.concat([to_long(punjab), to_long(haryana)], ignore_index=True)

print(f"Shape: {crop_data.shape}")
print(crop_data['state'].value_counts())
crop_data.head(6)
```

```python
# ── Save ──────────────────────────────────────────────────────────────
crop_data.to_csv('../Data/Processed/crop_production_clean.csv', index=False)
print(f"✅ Saved {len(crop_data)} rows → crop_production_clean.csv")
```
